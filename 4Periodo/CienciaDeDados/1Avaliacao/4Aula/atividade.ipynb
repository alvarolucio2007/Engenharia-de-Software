## Hipótese 1: Alguma cidade se destaca na compra de um produto em específico?

```python
pd.crosstab(df["Cidade"],df["Produto"],normalize="index")*100
```

- Belo Horizonte se destaca na compra de headsets.
- Curitiba, Notebooks.
- Fortaleza, Tablets.
- Manaus, Teclados.
- Porto Alegre, Smartphones.

## Hipótese 2: Existe relação entre a idade do cliente e tipo de produto comprado?

```python
idade_bins= pd.cut(df["Idade"], bins=[0, 25, 40, 60, 100], labels=["18-25", "26-40", "41-60", "60+"])
pd.crosstab(idade_bins, df["Produto"])
```

### Como visto acima,pessoas de 18-25 anos não compraram muitos produtos em geral, sendo superados por todos os outros grupos etários.

## Hipótese 3: Clientes mais velhos gastam mais dinheiro que clientes mais jovens?

```python
idade_bins=pd.cut(df["Idade"],bins=[0,25,40,60,100],labels=["18-25","26-40","41-60","60+"])
preco_normalizado=df["Preço"].astype(str).str.replace(",", ".").astype(float)
receita=preco_normalizado*df["Quantidade"]
receita.groupby(idade_bins).sum()
```

### Campeão absoluto: 41-60 anos de idade, gastando quase 3 milhões de reais, seguido por 26-40 anos, que gastaram 2,5.
