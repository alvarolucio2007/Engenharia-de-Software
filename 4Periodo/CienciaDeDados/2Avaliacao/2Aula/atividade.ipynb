```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('funcionarios_empresa.csv')
df.info()
```

```python
df_limpo=df
df_limpo.isnull().sum()
```

```python
df_limpo['Nivel_Educacao'] = df_limpo['Nivel_Educacao'].fillna(df_limpo['Nivel_Educacao'].mode()[0])
df_limpo['Avaliacao_Performance'] = df_limpo['Avaliacao_Performance'].fillna(df_limpo['Avaliacao_Performance'].mean())
df_limpo['Beneficios'] = df_limpo['Beneficios'].fillna(df_limpo['Beneficios'].mean())
df_limpo['Satisfacao_Trabalho'] = df_limpo['Satisfacao_Trabalho'].fillna(df_limpo['Satisfacao_Trabalho'].mean())
df_limpo.isnull().sum()
```
