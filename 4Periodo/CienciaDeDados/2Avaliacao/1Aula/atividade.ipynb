```python
dados=[15,20,21,20,36,15,25,15]
media=np.mean(dados)
mediana=np.median(dados)
moda=pd.Series(dados).mode()[0]
print(media,mediana,moda)
```

```python
dados = [12,15,18,22,25,27,30,35,40,80]
df=pd.DataFrame({'valores':dados})
amplitude= df['valores'].max() - df['valores'].min()
desvio_padrao = df['valores'].std()
iqr = df['valores'].quantile(0.75) - df['valores'].quantile(0.25)

print(amplitude,desvio_padrao,iqr)
```
