```python
df2=df.copy()
```

```python
colunas_texto = df2.select_dtypes(include=['str']).columns

df2[colunas_texto] = df2[colunas_texto].apply(lambda x: x.str.strip())
#A diferença em cada coluna categórica não mudou, já que eram todas diferentes, apenas foram retirados trailing spaces.
```

```python
colunas_num=df2.select_dtypes(include=['int','float','int64','float64']).columns
df2[colunas_num]=df2[colunas_num].replace(0,np.nan)
df2.isna().sum()
# 29 mil e 31 mil valors em capital-gain e capital-loss respectivamente
```

# 4

```python
colunas_num=df2.select_dtypes(include=['int','float','int64','float64']).columns
minmax_scaler=MinMaxScaler()
zscore_scaler=StandardScaler()
robust_scaler=RobustScaler()

df_minmax,df_zscore,df_robust=df2.copy(),df2.copy(),df2.copy()

df_minmax[colunas_num] = minmax_scaler.fit_transform(df2[colunas_num])
df_zscore[colunas_num] = zscore_scaler.fit_transform(df2[colunas_num])
df_robust[colunas_num] = robust_scaler.fit_transform(df2[colunas_num])

df_minmax[colunas_num].describe()
```

## Comparação de Resultados
- Z-score: Ajusta os dados para terem média igual a 0, e desvio padrão igual a 1, mas é extremamente sensível a valores extremos.
- Min-Max: Limita a faixa de valores entre 0 e 1, mas outliers forçam os limites de maneira que comprimem a maioria dos dados em uma faixa muito pequena, deixando a média, por exemplo, `0.13` em `capital-gain`.
- Robust: Demonstrou maior resiliência aos valores extremos, já que usa a mediana e interfalo interquartil (IQR), evitando que os outliers distorçam a distribuição dos dados.
