# 5
- Uma boa estratégia seria a remoção caso as linhas sejam exatamente iguais, pois são dados provavelmente corruptos vindos de um erro no sistema que deu os dados, e não se perderia nada pois ainda teria a cópia original.
- Caso algum `transaction_id` seja duplicado, a melhor estratégia seria transformar os IDs duplicados em novos IDs de compras, para garantir a integridade dos dados.

# Reflexão
- Depende principalmente do tipo de dados duplicados, como por exemplo, em nomes de produtos, se forem muito parecidos, podem ser duplicados. Contudo, IDs, já que sem duplicatas, podem ser apenas diferentes uns com os outros, não são repetidos.
- Não necessariamente, pois cada transação poderia ter um ID (UUID) único e seguro atrelado a ela, então uma transação duplicada seria tratada apenas como duas transações normais.
- Pois o CRISP-DM dita aprender com as regras da empresa, e colunas que definem o mesmo registro são necessárias pois podem apontar produtos que vendem melhor que outros e consumidores que compram mais que outros.
