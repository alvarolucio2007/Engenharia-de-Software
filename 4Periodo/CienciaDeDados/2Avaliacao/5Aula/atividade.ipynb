```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('dados.csv',sep=';')
df
```

df['Cidade'].unique()
#Quantidade original de cidades diferentes: 28
#Quantidade após normalização: 6

```python
#Método de normalização: transforma em lower case, remove espaços desnecessários, troca acento 'ã' e 'í' por 'a' e 'i' respectivamente, e troca espaços duplos por espaços únicos ('  ' por ' ')
df['Cidade']=df['Cidade'].str.lower().str.strip().str.replace('ã','a').str.replace('í','i').str.replace('  ',' ')
```

```python
df['Preço']
#Nota-se que o tipo é str ao invés de float. para isso, precisaremos fazer algumas mudanças.
```

```python
df['Preço']=pd.to_numeric(df['Preço'],errors='coerce')
df['Preço'].isnull().sum()
#74 nulos
```

```python
df['Data_Compra']=pd.to_datetime(df['Data_Compra'],dayfirst=True,errors='coerce')
df['Data_Compra'].max()
#Compra mais nova: 23 de agosto de 2026
```

```python
df['Data_Compra'].min()

#Compra mais antiga: 2 de janeiro de 2025
```
